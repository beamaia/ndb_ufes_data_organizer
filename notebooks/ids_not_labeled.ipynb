```python
import pathlib as pl
import pandas as pd
```

```python
LINKS_PATH = "data/ndb_ufes/link/images"
```

```python
images_linked = [int(file.name[7:11]) for file in pl.Path(LINKS_PATH).iterdir()]
```

```python
images_linked = sorted(images_linked)
```

```python
max_id = max(images_linked)
min_id = min(images_linked)
range_ids = set(range(min_id, max_id + 1))
```

```python
missing_ids = list(range_ids - set(images_linked))
```

```python
ids_df = pd.DataFrame({
        "missing_patch_id": missing_ids
    })
```

```python
ids_df
```

```python
ids_df["origin"] = -1
```

```python
ids_df
```

```python
ids_df.to_csv("missing_ids.csv", index=False)
```
